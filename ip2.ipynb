```python
import kagglehub
import pandas as pd
import numpy as np
import os
```

```python
path = kagglehub.dataset_download("carolzhangdc/imdb-5000-movie-dataset")

print("Path to dataset files:", path)
```

```python
df = pd.read_csv(os.path.join(path, "movie_metadata.csv"))

df.shape
```

```python
print("Broj potpuno dupliranih redova:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

# proveravamo koliko je redova ostalo
print("Dimenzije skupa:", df.shape)
print("Broj potpuno dupliranih redova:", df.duplicated().sum())
```

```python
# Zamenjujemo neprelomivi razmak običnim razmakom
# i uklanjamo praznine sa početka i kraja naziva.

df['movie_title'] = (df['movie_title'].str.replace('\xa0', ' ', regex=False).str.strip())
```

```python
df = df.drop_duplicates(subset=['movie_imdb_link', 'movie_title']) 

print(
    "Broj redova sa ponovljenim nazivom i linkom:",
    df.duplicated(subset=['movie_title', 'movie_imdb_link']).sum()
)
```

```python
# Broj filmova po zemlji
print(df['country'].value_counts(dropna=False).head(30))
```

```python
# Broj filmova i broj poznatih budžeta po zemlji

budget_by_country = df.groupby('country', dropna=False).agg(
    broj_filmova=('movie_title', 'count'),
    poznati_budzeti=('budget', 'count'),
    prosecni_budzet=('budget', 'mean'),
    medijana_budzeta=('budget', 'median'),
    maksimalni_budzet=('budget', 'max')
)

budget_by_country.sort_values(
    'broj_filmova',
    ascending=False
).round(2)
```

```python
def izbaci_po_trajanju(df, min_trajanje=40, max_trajanje=250):
    """Uklanja filmove kraće od min_trajanje i duže od max_trajanje minuta.
    Filmovi bez podatka o trajanju (NaN) se zadržavaju."""
    maska = df['duration'].between(min_trajanje, max_trajanje) | df['duration'].isna()

    izbaceni = df[~maska]
    print(f"Izbačeno filmova: {len(izbaceni)}")
    print(izbaceni[['movie_title', 'title_year', 'duration']].sort_values('duration'))

    return df[maska].copy()

df = izbaci_po_trajanju(df)
```

```python
df['imdb_id'] = df['movie_imdb_link'].str.extract(r'(tt\d+)')

# 1. uklanjanje unosa koji nisu (prikazani) filmovi ili nemaju podatke
za_brisanje = [
    'tt1571403',   # Harry Potter and the Deathly Hallows: Part I — video-igra
    'tt1680310',   # Harry Potter and the Deathly Hallows: Part II — video-igra
    'tt2983582',   # Destiny — video-igra
    'tt0490087',   # Black Water Transit — nedovršen, nikad prikazan
    'tt4607906',   # Romantic Schemer — trajanje nije dostupno
]
print("Broj redova pre:", len(df))
df = df[~df['imdb_id'].isin(za_brisanje)].copy()
print("Broj redova posle:", len(df))

# 2. ručno pronađeno trajanje (u minutima)
trajanje = {
    'tt1717210': 90,    # Should've Been Romeo
    'tt3099638': 145,   # Barfi (2013, kanada)
    'tt0249588': 153,   # Hum To Mohabbat Karega
    'tt1289419': 115,   # N-Secure
    'tt0442764': 144,   # Dil Jo Bhi Kahey...
    'tt4590482': 143,   # Karachi se Lahore
    'tt0381053': 110,   # The Naked Ape
}
for imdb_id, minuti in trajanje.items():
    df.loc[df['imdb_id'] == imdb_id, 'duration'] = minuti

print("Nedostaje trajanje:", df['duration'].isna().sum())
```
